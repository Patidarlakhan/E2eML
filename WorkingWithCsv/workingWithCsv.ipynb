```python
# Importing Pandas
import pandas as pd
```

```python
# Reading a local csv file
df = pd.read_csv('some_csv.csv')
```

```python
# Opening a csv file from url
import requests
from io import StringIO

url = ""
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

pd.read_csv(data)
```

```python
# Sep Parameter
pd.read_csv('movie_titles_metadata.tsv', sep='\t', names=['sno', 'name', 'release_year', 'rating', 'votes', 'genres'])
```

```python
# Index_col Paramater
pd.read_csv('some_csv.csv', index_col='enrole_id')
```

```python
# Header Parameter
pd.read_csv('some_csv.csv', header=1)
```

```python
# usecols parameter
pd.read_csv('some_csv.csv', usecols=['enrollee_id', 'gender'])
```

```python
# Sqeeze Parameter
pd.read_csv('some_csv.csv', usecols=['enrollee_id'], squeeze=True)
```

```python
# Skiprow/ nrow parameter
pd.read_csv('some_csv.csv', skiprows=[0, 1])
```

```python
# Encoding Parameter
pd.read_csv('some_csv.csv', encoding='latin-1')
```

```python
# Skip Bad Lines
pd.read_csv('some_csv.csv', encoding='latin-1', error_bad_lines=False)
```

```python
# dtypes Parameter
pd.read_csv('some_csv.csv', encoding='latin-1', error_bad_lines=False, dtype={'target':int})
```

```python
# Handling Dates
pd.read_csv('some_csv.csv', error_bad_lines=False, parse_dates=['date'])
```

```python
# Converters
def rename(name): 
    if name == 'Royal Challengers Banglore':
        return 'RCB'
    else:
        return name

pd.read_csv('some_csv.csv', error_bad_lines=False, parse_dates=['date'], converters={'team1':rename})
```

```python
# na_values Parameter
pd.read_csv('some_csv.csv', na_values=['-'])
```

```python
# Loading a huge dataset into chunks
dff = pd.read_csv('some_csv.csv', chunksize=5000)
```
